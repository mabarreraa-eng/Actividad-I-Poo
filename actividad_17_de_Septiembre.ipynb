```python
# Ejercicio 4

class Miembro:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def mostrar(self):
        print(f"{self.nombre}: {self.edad:.0f} años")


class Familia:
    def __init__(self, edad_juan):
        # Miembros de la familia
        self.juan = Miembro("Juan", edad_juan)
        self.alberto = Miembro("Alberto", (2 / 3) * edad_juan)
        self.ana = Miembro("Ana", (4 / 3) * edad_juan)

        # La edad de la mamá es la suma de las edades de los tres hijos
        edad_mama = (
            self.juan.edad
            + self.alberto.edad
            + self.ana.edad
        )

        self.mama = Miembro("Mamá", edad_mama)

    def mostrar_edades(self):
        print("\nEdades de la familia:")
        self.juan.mostrar()
        self.alberto.mostrar()
        self.ana.mostrar()
        self.mama.mostrar()


# Programa principal
print("EDADES DE LA FAMILIA")
print("________________________________")

edad_juan = float(input("Ingrese la edad de Juan: "))

familia = Familia(edad_juan)

familia.mostrar_edades()

```

```python
# Ejercicio numero 5
class Operacion:
    def __init__(self):
        self.suma = 0
        self.x = 20
        self.y = 0
    def calcular(self):
        self.suma = self.suma + self.x
        self.y = 40
        self.x = self.x + self.y ** 2
        self.suma = self.suma + self.x / self.y
    def mostrar_resultado(self):
        print("EL VALOR DE LA SUMA ES:", self.suma)
# Ejecucion del programa
operacion = Operacion()
operacion.calcular()
operacion.mostrar_resultado()
```

```python
# Ejercicio numero 12
class Empleado:
    def __init__(self, horas_trabajadas, valor_hora, porcentaje_retencion):
        self.horas_trabajadas = horas_trabajadas
        self.valor_hora = valor_hora
        self.porcentaje_retencion = porcentaje_retencion
    def calcular_salario_bruto(self):
        return self.horas_trabajadas * self.valor_hora
    def calcular_retencion(self):
        salario_bruto = self.calcular_salario_bruto()
        return salario_bruto * self.porcentaje_retencion
    def calcular_salario_neto(self):
        salario_bruto = self.calcular_salario_bruto()
        retencion = self.calcular_retencion()
        return salario_bruto - retencion
# Creacion del objeto
empleado = Empleado(48, 5000, 0.125)
# Obtenencion de resultados
salario_bruto = empleado.calcular_salario_bruto()
retencion = empleado.calcular_retencion()
salario_neto = empleado.calcular_salario_neto()
# Mostrar resultados
print("Salario bruto: $", salario_bruto)
print("Retención en la fuente: $", retencion)
print("Salario neto: $", salario_neto)
```

```python
# Ejercicio numero 14
class Numero:
    def __init__(self, valor):
        self.valor = valor
    def calcular_cuadrado(self):
        return self.valor ** 2
    def calcular_cubo(self):
        return self.valor ** 3
# Eleccion numérica
valor = float(input("Ingrese un número: "))
# Creacion de objeto
numero = Numero(valor)
# Mostrar resultados
print("Cuadrado:", numero.calcular_cuadrado())
print("Cubo:", numero.calcular_cubo())
```

```python
# Ejercicio numero 17
import math
class Circulo:
    def __init__(self, radio):
        self.radio = radio
    def calcular_area(self):
        return math.pi * self.radio ** 2
    def calcular_longitud(self):
        return 2 * math.pi * self.radio
# Ingrese el radio
radio = float(input("Ingrese el radio del círculo: "))
# Creacion del objeto
circulo = Circulo(radio)
# Mostrar resultados
print("Área del círculo:", circulo.calcular_area())
print("Longitud de la circunferencia:", circulo.calcular_longitud())
```
